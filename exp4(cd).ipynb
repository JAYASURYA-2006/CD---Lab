```python
%%writefile valvar.l

%{
#include "valvar.tab.h"
%}

%%
[a-zA-Z] { return LET; }
[0-9]    { return DIG; }
\n       { return 0; }
%%

int yywrap() {
    return 1;
}
```

```python
%%writefile valvar.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex(void);
int yyerror(char *s);
%}

%token LET DIG

%%
variable: var ;

var: var DIG
   | var LET
   | LET
   ;

%%

int main() {
    printf("Enter the variable:\n");
    yyparse();
    printf("Valid variable\n");
    return 0;
}

int yyerror(char *s) {
    printf("Invalid variable\n");
    exit(0);
}
```
